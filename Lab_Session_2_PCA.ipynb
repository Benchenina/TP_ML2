## Part A: PCA Implementation from Scratch

In this section:
1. Load dataset into Pandas DataFrame.
2. Display dimensions & descriptive statistics.
3. Standardize numerical variables using mean and standard deviation.
4. Compute the **Correlation Matrix** of the standardized data.
5. Calculate eigenvalues and eigenvectors of the correlation matrix using NumPy.
6. Sort eigenvalues in descending order and reorder eigenvectors.
7. Select the two eigenvectors associated with largest eigenvalues.
8. Project standardized data onto the two selected principal components.
9. Calculate explained variance ratio and cumulative explained variance.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# 1. Load the dataset into a Pandas DataFrame
data = {
    'Mathematics': [12, 15, 8, 17, 10, 14],
    'Statistics': [11, 14, 9, 16, 12, 15],
    'Programming': [14, 16, 10, 18, 11, 13],
    'Algorithms': [13, 15, 9, 17, 10, 14]
}
students = ['S1', 'S2', 'S3', 'S4', 'S5', 'S6']
df = pd.DataFrame(data, index=students)

print("--- Question 1: Dataset ---")
display(df)
```

```python
# 2. Display dimensions and descriptive statistics
print("--- Question 2: Dimensions & Descriptive Statistics ---")
print(f"Dataset Dimensions: {df.shape[0]} rows x {df.shape[1]} columns\n")
display(df.describe())
```

```python
#Standarization of the dataset (Younef)

X = df.values.astype(float)
mean = X.mean(axis=0)
std = X.std(axis=0, ddof=0)
X_std = (X - mean) / std
pd.DataFrame(X_std, index=df.index, columns=df.columns).round(4)
```

```python
# 3. Standardize numerical variables using mean and standard deviation
# StandardScaler scales features to mean = 0 and std = 1 (using N degrees of freedom)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df)

X_scaled_df = pd.DataFrame(X_scaled, index=students, columns=df.columns)
print("--- Question 3: Standardized Data ---")
display(X_scaled_df)
```

```python
# 4. Compute the Correlation Matrix of the standardized data
# (Calculated instead of covariance matrix as requested)
# Note: For standardized features, correlation matrix equals covariance matrix
corr_matrix = np.corrcoef(X_scaled, rowvar=False)

corr_df = pd.DataFrame(corr_matrix, index=df.columns, columns=df.columns)
print("--- Question 4: Correlation Matrix ---")
display(corr_df)
```

```python
# 5. Calculate eigenvalues and eigenvectors of the correlation matrix
eigenvalues, eigenvectors = np.linalg.eig(corr_matrix)

print("--- Question 5: Raw Eigenvalues & Eigenvectors ---")
print("Eigenvalues:\n", eigenvalues)
print("\nEigenvectors (columns):\n", eigenvectors)
```

```python
# 6. Sort eigenvalues in descending order and reorder eigenvectors
sorted_indices = np.argsort(eigenvalues)[::-1]
eigenvalues_sorted = eigenvalues[sorted_indices]
eigenvectors_sorted = eigenvectors[:, sorted_indices]

print("--- Question 6: Sorted Eigenvalues & Corresponding Eigenvectors ---")
print("Sorted Eigenvalues:\n", eigenvalues_sorted)
print("\nSorted Eigenvectors (columns correspond to PC1, PC2, PC3, PC4):\n", eigenvectors_sorted)
```

```python
# 7. Select the two eigenvectors associated with largest eigenvalues
top_2_eigenvectors = eigenvectors_sorted[:, :2]

print("--- Question 7: Top 2 Eigenvectors (Projection Matrix W) ---")
W_df = pd.DataFrame(top_2_eigenvectors, index=df.columns, columns=['Vector_PC1', 'Vector_PC2'])
display(W_df)
```

```python
# 8. Project the standardized data onto the two selected principal components
# Formula: Z = X_scaled @ W
X_projected = np.dot(X_scaled, top_2_eigenvectors)

projected_df = pd.DataFrame(X_projected, index=students, columns=['PC1', 'PC2'])
print("--- Question 8: Projected Data (From Scratch) ---")
display(projected_df)
```

```python
# 9. Calculate explained variance ratio and cumulative explained variance
explained_variance_ratio = eigenvalues_sorted / np.sum(eigenvalues_sorted)
cumulative_variance_ratio = np.cumsum(explained_variance_ratio)

var_summary = pd.DataFrame({
    'Eigenvalue': eigenvalues_sorted,
    'Explained Variance Ratio': explained_variance_ratio,
    'Cumulative Ratio': cumulative_variance_ratio,
    'Percentage (%)': explained_variance_ratio * 100,
    'Cumulative Percentage (%)': cumulative_variance_ratio * 100
}, index=[f'PC{i+1}' for i in range(len(eigenvalues_sorted))])

print("--- Question 9: Explained Variance Summary (From Scratch) ---")
display(var_summary)
```

## Part B: PCA using Scikit-Learn

In this section:
10. Apply PCA using predefined class from `sklearn.decomposition`.
11. Reduce dataset to two principal components.
12. Display transformed coordinates of students.
13. Calculate explained variance ratio and cumulative explained variance.


```python
# 10 & 11. Apply PCA using Scikit-learn and reduce to 2 components
pca_sklearn = PCA(n_components=2)
X_pca_sklearn = pca_sklearn.fit_transform(X_scaled)

# 12. Display transformed coordinates of the students
sklearn_coords_df = pd.DataFrame(X_pca_sklearn, index=students, columns=['PC1_sklearn', 'PC2_sklearn'])
print("--- Question 12: Transformed Coordinates (Scikit-Learn) ---")
display(sklearn_coords_df)
```

```python
# 13. Calculate explained variance ratio and cumulative explained variance
sklearn_var_ratio = pca_sklearn.explained_variance_ratio_
sklearn_cum_ratio = np.cumsum(sklearn_var_ratio)

sklearn_summary = pd.DataFrame({
    'Explained Variance Ratio': sklearn_var_ratio,
    'Cumulative Ratio': sklearn_cum_ratio,
    'Percentage (%)': sklearn_var_ratio * 100,
    'Cumulative Percentage (%)': sklearn_cum_ratio * 100
}, index=['PC1', 'PC2'])

print("--- Question 13: Explained Variance (Scikit-Learn) ---")
display(sklearn_summary)
```

### Interpretation & Theoretical Answers

#### Question 17: Explain any differences observed between the results, particularly possible sign changes in the principal components.
1. **Sign Indeterminacy**:
   - An eigenvector $\mathbf{v}$ satisfies $A\mathbf{v} = \lambda\mathbf{v}$.
   - Negating both sides yields $A(-\mathbf{v}) = \lambda(-\mathbf{v})$.
   - Therefore, both $\mathbf{v}$ and $-\mathbf{v}$ are valid eigenvectors representing the exact same line/subspace.
2. **Implementation Difference**:
   - `np.linalg.eig` solves the characteristic equation of the covariance/correlation matrix.
   - `sklearn.decomposition.PCA` uses Singular Value Decomposition (SVD) with deterministic sign-flipping conventions (`svd_flip`).
3. **Geometric Impact**:
   - A sign difference merely reflects the projected points across that axis (mirror image).
   - The relative distances between points, clusters, variances, and all geometric relationships are completely preserved.
